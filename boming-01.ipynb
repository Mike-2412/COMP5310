```python
import pandas as pd
from IPython.display import display

train = pd.read_csv("train_occupancy.csv")
air = pd.read_csv("airline_delay.csv")
hotel = pd.read_csv("hotel_bookings.csv")

display(train.head())
display(air.head())
display(hotel.head())
```

```python
comparison = pd.DataFrame({
    "dataset": [
        "Train occupancy",
        "Air travel delay",
        "Hotel booking"
    ],

    "rows": [
        len(train),
        len(air),
        len(hotel)
    ],

    "columns": [
        train.shape[1],
        air.shape[1],
        hotel.shape[1]
    ],

    "numeric_columns": [
        len(train.select_dtypes(include="number").columns),
        len(air.select_dtypes(include="number").columns),
        len(hotel.select_dtypes(include="number").columns)
    ],

    "columns_with_missing": [
        train.isna().any().sum(),
        air.isna().any().sum(),
        hotel.isna().any().sum()
    ],

    "rows_with_missing_pct": [
        train.isna().any(axis=1).mean() * 100,
        air.isna().any(axis=1).mean() * 100,
        hotel.isna().any(axis=1).mean() * 100
    ],

    "exact_duplicate_rows": [
        train.duplicated().sum(),
        air.duplicated().sum(),
        hotel.duplicated().sum()
    ],

    "exact_duplicate_pct": [
        train.duplicated().mean() * 100,
        air.duplicated().mean() * 100,
        hotel.duplicated().mean() * 100
    ]
})

comparison["other_columns"] = (
    comparison["columns"] - comparison["numeric_columns"]
)

display(comparison.round(2))
```

```python
train_audit = pd.DataFrame({
    "read_type": train.dtypes.astype(str),
    "missing_count": train.isna().sum(),
    "missing_pct": train.isna().mean() * 100,
    "unique_nonmissing": train.nunique()
})

train_audit = train_audit.sort_values(
    "missing_pct",
    ascending=False
)

display(train_audit.round(2))

air_audit = pd.DataFrame({
    "read_type": air.dtypes.astype(str),
    "missing_count": air.isna().sum(),
    "missing_pct": air.isna().mean() * 100,
    "unique_nonmissing": air.nunique()
})

air_audit = air_audit.sort_values(
    "missing_pct",
    ascending=False
)

display(air_audit.round(2))

hotel_audit = pd.DataFrame({
    "read_type": hotel.dtypes.astype(str),
    "missing_count": hotel.isna().sum(),
    "missing_pct": hotel.isna().mean() * 100,
    "unique_nonmissing": hotel.nunique()
})

hotel_audit = hotel_audit.sort_values(
    "missing_pct",
    ascending=False
)

display(hotel_audit.round(2))
```

```python
print("train:")
display(
    train["Segment.Direction"].value_counts(dropna=False)
)

print("Occupancy status:")
display(
    train["Occupancy Status"].value_counts(dropna=False)
)

print("Occupancy range:")
display(
    train["Occupancy Range"].value_counts(dropna=False)
)

print("Occupancy range percentage (%):")
display(
    train["Occupancy Range"]
    .value_counts(dropna=False, normalize=True)
    .mul(100)
    .round(2)
)
```
